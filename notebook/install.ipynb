# 在 macbook 上安装

## 安装必要的组件

### Deno

先查看 .tool-versions 文件中约定的 deno 的版本号，然后安装 deno。

```bash
wget https://deno.land/install.sh
sh install.sh v1.43.5
```

如果网络可以，可以直接执行下面的命令：

```bash
curl -fsSL https://deno.land/install.sh | sh -s v1.43.5
```


### Meteor

先查需要使用哪个版本，然后直接下载安装，如果网络不好，多试几次。

```bash

cat apps/meteor/.meteor/release

METEOR@3.3

curl https://install.meteor.com/\?release\=3.3 | sh

meteor --version

```

### Run the Rocket.Chat project

```bash

yarn
yarn build

# 大于 16G 用下面命令
yarn dev

# 小于 16G 用下面命令
yarn dsv

```
